```python

# Reload the original datasets to avoid repeated merging
movies = pd.read_csv("dataset/tmdb_5000_movies.csv")
credits = pd.read_csv("dataset/tmdb_5000_credits.csv")

# Select only the required columns from credits
credits = credits[["movie_id", "cast", "crew"]]

# Merge both datasets
movies = movies.merge(
    credits,
    left_on="id",
    right_on="movie_id"
)

# Check the result
print("Combined Dataset Shape:", movies.shape)

display(movies[["title", "genres", "cast", "crew"]].head())
```

```python

# Check missing values in important columns

important_columns = [
    "title",
    "genres",
    "keywords",
    "overview",
    "cast",
    "crew"
]

print("Missing values in important columns:\n")

print(movies[important_columns].isnull().sum())
```

```python
# Replace missing movie overviews with empty strings
movies["overview"] = movies["overview"].fillna("")

# Verify that there are no missing overviews
print("Missing overviews after cleaning:",
      movies["overview"].isnull().sum())
```

```python
import json

# Convert JSON text into a Python list of dictionaries
def convert_to_list(text):
    try:
        return json.loads(text)
    except:
        return []

# Extract names from genres, keywords and cast
def extract_names(text):
    return [item["name"] for item in convert_to_list(text)]

# Extract the director's name from crew information
def extract_director(text):
    for person in convert_to_list(text):
        if person.get("job") == "Director":
            return person["name"]
    return ""

# Apply the functions to our dataset
movies["genres"] = movies["genres"].apply(extract_names)
movies["keywords"] = movies["keywords"].apply(extract_names)
movies["cast"] = movies["cast"].apply(extract_names)
movies["director"] = movies["crew"].apply(extract_director)

# Display the result
display(movies[["title", "genres", "keywords", "cast", "director"]].head())
```

```python

# Convert lists into text so we can combine them

movies["genres"] = movies["genres"].apply(lambda x: " ".join(x))
movies["keywords"] = movies["keywords"].apply(lambda x: " ".join(x))
movies["cast"] = movies["cast"].apply(lambda x: " ".join(x))

# Combine all important movie information into one column


# Combine all important movie information
movies["tags"] = (
    movies[["genres", "keywords", "overview", "actors", "director"]]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
)

# Display the result
display(movies[["title", "tags"]].head())

```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

vectors = tfidf.fit_transform(movies["tags"])

print("Vector matrix shape:", vectors.shape)
```

```python

from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between every pair of movies

similarity = cosine_similarity(vectors)

# Check the result

print("Similarity matrix shape:", similarity.shape)

# Check how similar Avatar is to itself

print("Avatar similarity with itself:", similarity[0][0])
```

```python
# Function to recommend similar movies
def recommend_movies(movie_title, number_of_recommendations=5):

    # Find the index of the selected movie
    movie_index = movies[movies["title"] == movie_title].index[0]

    # Get similarity scores for the selected movie
    scores = list(enumerate(similarity[movie_index]))

    # Sort movies from most similar to least similar
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Remove the selected movie itself
    scores = scores[1:number_of_recommendations + 1]

    # Get the titles of recommended movies
    recommended_movies = []

    for index, score in scores:
        recommended_movies.append(movies.iloc[index]["title"])

    return recommended_movies


# Test our recommendation system
print("Movies similar to Avatar:")
print(recommend_movies("Avatar"))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

# Convert movie tags into TF-IDF vectors
vectors_tfidf = tfidf.fit_transform(movies["tags"])

# Calculate similarity using the new vectors
similarity_tfidf = cosine_similarity(vectors_tfidf)

print("TF-IDF Vector Shape:", vectors_tfidf.shape)
print("New Similarity Matrix Shape:", similarity_tfidf.shape)
```

```python
# Find The Dark Knight's index
dark_knight_index = movies[
    movies["title"] == "The Dark Knight"
].index[0]

# Get TF-IDF similarity scores
scores = list(enumerate(similarity_tfidf[dark_knight_index]))

# Sort from highest to lowest similarity
scores = sorted(scores, key=lambda x: x[1], reverse=True)

# Display the top 5 recommendations
print("TF-IDF Recommendations for The Dark Knight:\n")

for index, score in scores[1:6]:
    print(movies.iloc[index]["title"], ":", round(score, 4))
```

```python
def recommend_movies_tfidf(movie_title, number_of_recommendations=5):

    # Find the selected movie
    movie_index = movies[
        movies["title"].str.lower() == movie_title.lower()
    ].index

    # Check whether the movie exists
    if len(movie_index) == 0:
        return "Movie not found in dataset."

    movie_index = movie_index[0]

    # Get similarity scores
    scores = list(enumerate(similarity_tfidf[movie_index]))

    # Sort by similarity
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Remove the selected movie
    scores = scores[1:number_of_recommendations + 1]

    # Collect recommended movie titles
    recommendations = []

    for index, score in scores:
        recommendations.append({
            "Movie": movies.iloc[index]["title"],
            "Similarity Score": round(score, 4)
        })

    return pd.DataFrame(recommendations)
```

```python
# Convert the stored text into Python lists
cast_data = ast.literal_eval(sample["cast"])
crew_data = ast.literal_eval(sample["crew"])

# Extract the first 5 actors
actors = [person["name"] for person in cast_data[:5]]

# Extract the director
directors = [
    person["name"]
    for person in crew_data
    if person.get("job") == "Director"
]

print("Movie:", sample["title"])
print("\nTop 5 Actors:", actors)
print("\nDirector:", directors)
```

```python
# Function to extract the first 5 actors
def extract_actors(data):
    try:
        people = ast.literal_eval(data)
        return " ".join(person["name"] for person in people[:5])
    except:
        return ""

# Function to extract the director
def extract_director(data):
    try:
        people = ast.literal_eval(data)

        for person in people:
            if person.get("job") == "Director":
                return person["name"]

        return ""
    except:
        return ""

# Create new columns
movies["actors"] = movies["cast"].apply(extract_actors)
movies["director"] = movies["crew"].apply(extract_director)

print("New columns created successfully!")

print(movies[["title", "actors", "director"]].head(10))
```

```python
def get_recommendations(movie_title, top_n=5):

    # Find the selected movie
    matches = movies.index[
        movies["title"].str.lower() == movie_title.lower()
    ].tolist()

    if not matches:
        return "Movie not found!"

    movie_index = matches[0]

    # Get similarity scores
    scores = list(enumerate(similarity[movie_index]))

    # Sort movies by similarity
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Remove the selected movie itself
    scores = [item for item in scores if item[0] != movie_index]

    # Get the top recommended movies
    movie_indices = [item[0] for item in scores[:top_n]]

    return movies.iloc[movie_indices][
        ["title", "director", "actors"]
    ]
```
